### (7): Using a 2-sample t-test**, determine if there is a statistical difference between the price of diamonds with a cut that is considered:
##### 1. 'Good' vs 'Very Good'
##### 2. 'Premium' vs 'Ideal'
##### **set alpha to 5% i.e. 0.05.

##### Two Sample T-Test for determining if there is a statistical difference between the price of diamonds with good cut and those with very good cut

```python
#two sample t-test: is there a difference in the price of good cut diamonds and very good cut diamonds

#extract the data for good cut diamonds and very good cut diamonds
sample_good_cut   = diamonds_df[diamonds_df['cut'] == 'Good']['price']
sample_very_good_cut = diamonds_df[diamonds_df['cut'] == 'Very Good']['price']


alpha = 0.05
t_stat, p_value = stats.ttest_ind(sample_good_cut, sample_very_good_cut) #perform the t-test
print('p-value: ' + str(p_value))

#compare the p_value with alpha to determine the significance
if(p_value < alpha):
    print('The price of good-cut diamonds and very-good-cut diamonds are statistically different.')
else:
    print('The price of good-cut diamonds and very-good-cut diamonds are not statistically different.')
```

Since the calculated p value is greater than alpha 0.05, we can conclude that the price of good-cut diamonds and very-good-cut diamonds are not statistically different.

##### Two Sample T-Test for determining if there is a statistical difference between the price of diamonds with premium cut and those with ideal cut

```python
#two sample t-test: is there a difference in the price of good cut diamonds and very good cut diamonds

#extract the data for good cut diamonds and very good cut diamonds
sample_premium_cut   = diamonds_df[diamonds_df['cut'] == 'Premium']['price']
sample_ideal_cut = diamonds_df[diamonds_df['cut'] == 'Ideal']['price']


alpha1 = 0.05
t_stat1, p_value1 = stats.ttest_ind(sample_premium_cut, sample_ideal_cut) #perform the t-test
print('p-value: ' + str(p_value1))

#compare the p_value with alpha to determine the significance
if(p_value1 < alpha1):
    print('The price of premium-cut diamonds and ideal-cut diamonds are statistically different.')
else:
    print('The price of premium-cut diamonds and ideal-cut diamonds are not statistically different.')
```

Since the calculated p value is smaller than alpha 0.05, we can conclude that the price of premium-cut diamonds and ideal-cut diamonds are statistically different.
