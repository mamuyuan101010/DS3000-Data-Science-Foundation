```python
#Import libraries
import numpy as np
```

(1):

```python
#Load exam_grades.csv data into a numpy array
data = np.loadtxt('exam_grades.csv', #the file name
                  delimiter=',', #indicate the delimiter
                  skiprows=1) #skip the header

print(data)
```

(2):

```python
# numbers of observations and columns in the data set
print('There are ' + str(data.shape[0]) + ' observations in the dataset')
print('There are ' + str(data.shape[1]) + ' columes in the dataset')
```

(3):

```python
# minimum and maximum overall course grade for the statistics course
min = round(data[:,4].min(),2)
max = round(data[:,4].max(),2)
print('The minimum overall course grade for the statistics course is ' + str(min) + '.')
print('The maximum overall course grade for the statistics course is ' + str(max) + '.')
```

(4):

```python
# remove duplicate years and list year in rising order
years_covered = np.unique(data[:,0])
years_covered.sort()
print(str(years_covered))
```

(5):

```python
# count the number of students who took the course each year
stu_2000 = data[data[:,0] == 2000]
stu_2001 = data[data[:,0] == 2001]
stu_2002 = data[data[:,0] == 2002]
stu_2003 = data[data[:,0] == 2003]
print('There are ' + str(stu_2000.shape[0]) + ' students in 2000.')
print('There are ' + str(stu_2001.shape[0]) + ' students in 2001.')
print('There are ' + str(stu_2002.shape[0]) + ' students in 2002.')
print('There are ' + str(stu_2003.shape[0]) + ' students in 2003.')
```

(6):

```python
# calculate the minimum, maximum, mean and standard deviation 
# for each of the three exam scores. Summarize your findings
# each year
exam1 = data[:,1]
exam2 = data[:,2]
exam3 = data[:,3]
print('Exam 1 has the following statistic value:')
print('min:' + str(round(exam1.min(),2)))
print('max:' + str(round(exam1.max(),2)))
print('mean:' + str(round(exam1.mean(),2)))
print('standard deviation:' + str(round(exam1.std(),2)))
print()
print('Exam 2 has the following statistic value:')
print('min:' + str(round(exam2.min(),2)))
print('max:' + str(round(exam2.max(),2)))
print('mean:' + str(round(exam2.mean(),2)))
print('standard deviation:' + str(round(exam2.std(),2)))
print()
print('Exam 3 has the following statistic value:')
print('min:' + str(round(exam3.min(),2)))
print('max:' + str(round(exam3.max(),2)))
print('mean:' + str(round(exam3.mean(),2)))
print('standard deviation:' + str(round(exam3.std(),2)))
```
